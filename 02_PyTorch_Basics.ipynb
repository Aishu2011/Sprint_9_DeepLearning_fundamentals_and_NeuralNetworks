### Definition

**Slicing** selects a range or subset of tensor elements.

# 13. Tensor Reshaping

Sometimes the same data needs to be arranged into a different shape without changing the underlying number of elements.

For example:

`2 × 6 = 12 elements`

can be reshaped into:

`3 × 4 = 12 elements`

The number of elements must remain compatible.

### Real-world analogy

Imagine 12 books arranged in 2 shelves with 6 books each.

I can rearrange them into 3 shelves with 4 books each without creating or removing books.

### AI/ML usage

Reshaping is common when moving data between:

- Layers
- Batches
- Image representations
- Feature matrices

````python
x = torch.arange(12)

print("Original:", x)
print("Original shape:", x.shape)

reshaped = x.reshape(3, 4)

print("Reshaped:")
print(reshaped)
print("New shape:", reshaped.shape)
````

### Result

The values remain the same, but their arrangement changes from one dimension to a `3 × 4` matrix.

This is useful when a neural-network layer expects a particular input shape.

# 14. `reshape()`

# 15. `view()`

# 18. Matrix Multiplication

### Definition

**Matrix multiplication** combines rows from one matrix with columns from another matrix.

In PyTorch, matrix multiplication can be performed using:

- `torch.matmul()`
- `@`

### Why it matters

Matrix multiplication is one of the most important operations in neural networks.

A linear layer essentially performs a matrix multiplication followed by adding a bias.

### Real-world analogy

Imagine each row contains a customer's feature values and each column represents a set of learned weights.

Multiplication combines the information to calculate new values.

### AI/ML usage

Matrix multiplication is used in:

- Linear layers
- Neural-network forward passes
- Attention calculations
- Embedding operations
- Many optimization calculations

````python
A = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

B = torch.tensor([
    [5.0, 6.0],
    [7.0, 8.0]
])

C = A @ B

print("A:")
print(A)

print("B:")
print(B)

print("A @ B:")
print(C)
````

### Result explanation

The result is not element-by-element multiplication.

Each output value comes from the dot product of one row from `A` with one column from `B`.

This is why matrix dimensions must be compatible.

For a product:

**(m × n) @ (n × p) → (m × p)**

The middle dimensions must match.

# 19. Broadcasting

### Definition

**Broadcasting** allows PyTorch to perform operations between tensors with compatible but different shapes.

PyTorch automatically expands the smaller tensor conceptually to make the operation possible.

### Real-world analogy

Imagine adding the same fixed adjustment to every row of a table.

Instead of manually repeating the adjustment for every row, broadcasting lets the framework apply it automatically.

### AI/ML usage

Broadcasting is useful for:

- Adding bias
- Normalization
- Feature transformations
- Batch operations

````python
data = torch.tensor([
    [10.0, 20.0, 30.0],
    [40.0, 50.0, 60.0]
])

adjustment = torch.tensor([1.0, 2.0, 3.0])

result = data + adjustment

print("Data:")
print(data)

print("Adjustment:", adjustment)

print("Result:")
print(result)
````

### Result explanation

The adjustment tensor has shape `(3,)`.

PyTorch applies it to each row of the `2 × 3` tensor:

- first column + 1
- second column + 2
- third column + 3

This avoids manually copying the adjustment for every row.

# 27. `requires_grad`

### Definition

`requires_grad=True` tells PyTorch that I want to track operations involving a tensor so that gradients can later be calculated.

Gradients tell us how much a value needs to change to reduce a loss.

This is a key part of neural-network training.

### Real-world analogy

Imagine adjusting the temperature of an oven.

If I know how the final result changes when I slightly change the temperature, I can use that information to improve the setting.

The gradient provides this direction/sensitivity information.

### AI/ML usage

Gradients are used to update model parameters during training.

Typical flow:

**Prediction → Loss → Gradient → Parameter update**

````python
x = torch.tensor(3.0, requires_grad=True)

y = x**2

y.backward()

print("x:", x)
print("y:", y)
print("Gradient of y with respect to x:", x.grad)
````

### Result explanation

The calculation is:

**y = x²**

For `x = 3`, the derivative is:

**dy/dx = 2x = 6**

PyTorch calculates this automatically after:

`y.backward()`

So `x.grad` contains the gradient `6`.

This is the basic idea behind **automatic differentiation**.

# 28. Why Gradients Matter in Neural Networks

Suppose a neural network makes a prediction that is not accurate.

The model calculates a **loss**, which measures the error.

Then PyTorch calculates gradients showing how changing the parameters would affect that loss.

An optimizer uses those gradients to update the parameters.

### Simplified training loop

**Input**

→ **Model**

→ **Prediction**

→ **Loss**

→ **Backward / gradients**

→ **Optimizer update**

→ **Repeat**

This is why `requires_grad` and autograd are so important in PyTorch.

# 29. `torch.no_grad()`

### Definition

`torch.no_grad()` temporarily disables gradient tracking inside its block.

### Why use it?

During inference/evaluation, I usually do not need gradients.

Disabling gradient tracking:

- reduces unnecessary computation
- reduces memory usage
- makes inference more efficient

### Real-world analogy

During practice, a student may record every learning step.

During the final exam, they only need to produce answers. They do not need to record all the learning calculations.

### AI/ML usage

`torch.no_grad()` is commonly used during model inference and evaluation.

````python
x = torch.tensor(3.0, requires_grad=True)

with torch.no_grad():
    y = x**2

print("x requires grad:", x.requires_grad)
print("y requires grad:", y.requires_grad)
````

### Result explanation

`x` still has `requires_grad=True`.

But the calculation performed inside `torch.no_grad()` does not build the normal gradient-tracking graph, so `y` does not require gradients.

This is useful when the trained model is only being used to make predictions.

# 30. `requires_grad` vs `torch.no_grad()`

| Concept | Purpose |
|---|---|
| `requires_grad=True` | Track operations so gradients can be calculated |
| `y.backward()` | Calculate gradients through the computation graph |
| `torch.no_grad()` | Disable gradient tracking temporarily |
| `.grad` | Stores the calculated gradient for a tracked tensor |
