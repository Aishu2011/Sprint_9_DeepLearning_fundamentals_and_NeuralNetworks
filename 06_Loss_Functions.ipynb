# Notebook 6 – Loss Functions

## 1. What is a Loss Function?

A **loss function** is a mathematical function that measures the difference between a model's prediction and the actual target.

### In my own words

I think of loss as a **mistake meter**.

- Small loss → prediction is close to the actual value.
- Large loss → prediction is far from the actual value.

During neural-network training, the model tries to reduce this loss.

### Real-world example

Suppose a model predicts the price of a product as ₹950, but the actual price is ₹1000.

The loss function converts this difference into a numerical value. The training process uses that value to learn better parameters.

### AI/ML usage

Loss functions are used during model training for tasks such as:
- predicting prices
- predicting sales
- detecting spam
- predicting whether a customer will leave
- classifying images into multiple classes

## 2. Prediction → Loss → Learning

The basic training idea is:

**Input → Model → Prediction → Loss → Parameter Update → Better Prediction**

For example:

1. The input is given to the neural network.
2. The network produces a prediction.
3. The loss function compares prediction and actual target.
4. A large loss means the prediction needs more correction.
5. Backpropagation calculates gradients.
6. The optimizer updates weights and biases.
7. The process repeats.

### Important distinction

A **loss function** measures error for training.

A **metric** is usually used to evaluate model performance in a way that is easier to interpret, such as accuracy or MAE.

## 3. Mean Absolute Error (MAE)

### Definition

**MAE = Mean Absolute Error**

It calculates the average absolute difference between actual values and predictions.

### Formula

MAE = (1/n) × Σ |y - ŷ|

Where:
- `y` = actual value
- `ŷ` = predicted value
- `n` = number of observations

### Mathematical intuition

Absolute value removes the negative sign.

So an error of `-10` and an error of `+10` both contribute `10`.

### Real-world example

If a delivery-time model predicts:

Actual: `[30, 50, 40]` minutes  
Prediction: `[35, 45, 42]` minutes

The errors are `5, 5, 2`.

MAE tells us the average size of these errors.

### When to use

MAE is useful when:
- you want an error measure in the same unit as the target
- you do not want very large errors to dominate as strongly as MSE
- the data may contain some outliers

### When not to use

MAE may not be the best choice when:
- very large errors must be strongly penalized
- you specifically want a smooth squared-error objective

### Advantages

- Easy to understand
- Same unit as the target
- More resistant to outliers than MSE

### Limitations

- The absolute-value function is not differentiable exactly at zero
- It penalizes all error sizes linearly

### MAE – Simple numerical example

Actual values:

`[10, 20, 30]`

Predicted values:

`[12, 18, 33]`

Absolute errors:

- `|10 - 12| = 2`
- `|20 - 18| = 2`
- `|30 - 33| = 3`

Therefore:

MAE = `(2 + 2 + 3) / 3 = 2.33`

So, on average, the prediction is about **2.33 units away** from the actual value.

```python
import numpy as np

actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 33])

mae = np.mean(np.abs(actual - predicted))
print("MAE:", mae)
```

```python
import torch

actual_t = torch.tensor([10.0, 20.0, 30.0])
predicted_t = torch.tensor([12.0, 18.0, 33.0])

mae_t = torch.mean(torch.abs(actual_t - predicted_t))
print("PyTorch MAE:", mae_t.item())
```

### Interpretation

The result should be approximately **2.33**.

This means the average absolute prediction error is about 2.33 target units.

## 4. Mean Squared Error (MSE)

### Definition

**MSE = Mean Squared Error**

It calculates the average of the squared differences between actual and predicted values.

### Formula

MSE = (1/n) × Σ(y - ŷ)²

### Mathematical intuition

The error is squared before averaging.

This makes large errors much more important.

For example:

- error = 2 → squared error = 4
- error = 10 → squared error = 100

So an error of 10 has much more influence than an error of 2.

### Real-world example

For house-price prediction, a prediction that is off by ₹100,000 may be much more serious than one that is off by ₹10,000. MSE gives the larger error much more weight.

### When to use

Use MSE when:
- large errors should be strongly penalized
- the target is continuous
- squared-error optimization is appropriate

### When not to use

MSE may be less suitable when:
- the dataset contains strong outliers
- you want an error measure in the original target unit

### Advantages

- Smooth and differentiable
- Strongly penalizes large errors
- Commonly used in regression

### Limitations

- Sensitive to outliers
- Units become squared units

### MSE – Simple numerical example

Actual:

`[10, 20, 30]`

Predicted:

`[12, 18, 33]`

Squared errors:

- `(10 - 12)² = 4`
- `(20 - 18)² = 4`
- `(30 - 33)² = 9`

MSE = `(4 + 4 + 9) / 3 = 5.67`

The larger error of 3 contributes 9 because it is squared.

```python
actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 33])

mse = np.mean((actual - predicted) ** 2)
print("MSE:", mse)
```

```python
mse_t = torch.mean((actual_t - predicted_t) ** 2)
print("PyTorch MSE:", mse_t.item())
```

### Interpretation

The result should be approximately **5.67**.

MSE is larger than MAE here because the errors are squared.

## 5. Root Mean Squared Error (RMSE)

### Definition

**RMSE = Root Mean Squared Error**

RMSE is the square root of MSE.

### Formula

RMSE = √[(1/n) × Σ(y - ŷ)²]

### Mathematical intuition

RMSE keeps the MSE behavior of strongly penalizing large errors, but the final square root brings the result back to the original target unit.

### Real-world example

If a model predicts house prices in rupees, RMSE is also expressed in rupees.

### When to use

Use RMSE when:
- large errors should matter
- you want the final value in the same unit as the target
- you want a common regression error measure

### When not to use

RMSE may not be ideal when:
- outliers should have little influence
- you prefer the more robust behavior of MAE

### Advantages

- Same unit as the target
- Penalizes large errors
- Easy to compare with the scale of the target

### Limitations

- Sensitive to outliers
- More affected by large mistakes than MAE

### RMSE – Simple numerical example

From the previous example:

MSE = `5.67`

Therefore:

RMSE = `√5.67 ≈ 2.38`

So the typical error is about **2.38 target units**, with larger errors receiving extra penalty.

```python
rmse = np.sqrt(mse)
print("RMSE:", rmse)
```

```python
rmse_t = torch.sqrt(mse_t)
print("PyTorch RMSE:", rmse_t.item())
```

## 6. MAE vs MSE vs RMSE

| Loss | Main idea | Outlier sensitivity | Same unit as target? |
|---|---|---|---|
| MAE | Average absolute error | Lower | Yes |
| MSE | Average squared error | High | No |
| RMSE | Square root of MSE | High | Yes |

### Easy memory trick

- **MAE** → how far away, on average
- **MSE** → large errors get amplified
- **RMSE** → amplified errors, but converted back to original units

# Part B – Use the Supplied Dataset

The supplied dataset is the Online Retail dataset. I will create a simple regression-style target from the available numerical columns.

For this demonstration:
- `Quantity` is used as one input.
- `UnitPrice` is used as another input.
- `TotalAmount = Quantity × UnitPrice` is treated as a continuous target.

This is only a learning example. The purpose is to understand how regression losses behave on real data.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv", encoding="latin1")

df = df[["Quantity", "UnitPrice"]].copy()
df = df.replace([np.inf, -np.inf], np.nan).dropna()
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]

df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]

df.head()
```

- RMSE converts the squared-error result back to the target's unit.

The numerical values depend on the selected rows from the supplied dataset.

# Part C – Binary Classification

Binary classification has two possible classes.

Examples:
- spam / not spam
- fraud / not fraud
- customer leaves / customer stays
- disease / no disease

The model often produces a probability between 0 and 1 for the positive class.

## 7. Binary Cross Entropy (BCE)

### Definition

Binary Cross Entropy measures the difference between the true binary label and the predicted probability.

For one example:

BCE = -[y log(p) + (1-y) log(1-p)]

Where:
- `y` = actual label, 0 or 1
- `p` = predicted probability of class 1

### Mathematical intuition

If the true class is 1:
- predicting probability 0.99 → very small loss
- predicting probability 0.10 → large loss

If the true class is 0:
- predicting probability 0.01 → very small loss
- predicting probability 0.90 → large loss

So BCE strongly discourages confident wrong predictions.

### Real-world example

For spam detection:
- actual = spam
- model probability of spam = 0.95

This should have a small loss.

If the model gives spam probability = 0.05, the loss should be large.

### When to use

Use BCE for binary classification when the model output is a probability between 0 and 1.

### When not to use

Do not directly use plain BCE on raw logits. For raw logits, use a logits-aware loss such as `BCEWithLogitsLoss`.

### Advantages

- Designed for binary probabilities
- Penalizes confident wrong predictions strongly
- Works naturally with probabilistic binary classification

### Limitations

- Predictions must be valid probabilities when using plain BCE
- Numerical stability requires care near probabilities 0 and 1

### BCE – Simple numerical example

Suppose:

Actual label = `1`

Predicted probability = `0.8`

Then:

BCE = `-[1 × log(0.8) + 0 × log(0.2)]`

So:

BCE = `-log(0.8) ≈ 0.223`

A prediction of 0.8 for a true positive therefore produces a relatively small loss.

```python
y_true = 1.0
p = 0.8

bce = -(y_true * np.log(p) + (1 - y_true) * np.log(1 - p))
print("BCE:", bce)
```

```python
y_true_t = torch.tensor([1.0])
p_t = torch.tensor([0.8])

bce_t = torch.nn.functional.binary_cross_entropy(p_t, y_true_t)
print("PyTorch BCE:", bce_t.item())
```

### Interpretation

The loss is approximately **0.223**.

If the prediction becomes closer to 1 for this true-positive example, the BCE becomes smaller.

## 8. BCE with Logits – Concept

### What is a logit?

A **logit** is the raw output of a binary classifier before the sigmoid function.

The sigmoid converts a logit into a probability:

p = 1 / (1 + e^-z)

where `z` is the logit.

### Why use BCEWithLogitsLoss?

Instead of manually doing:

**logit → sigmoid → BCE**

PyTorch provides:

**BCEWithLogitsLoss(logit, target)**

It combines the sigmoid operation and BCE calculation in a numerically stable way.

### Simple intuition

Think of the model as producing a raw confidence score first.

- Large positive logit → probability close to 1
- Large negative logit → probability close to 0
- Logit near 0 → probability near 0.5

### AI/ML usage

This is commonly used for neural networks performing binary classification.

### When to use

Use `BCEWithLogitsLoss` when your model's final layer produces **raw logits** and the target is binary.

### Important practical point

If using `BCEWithLogitsLoss`, do **not** add a sigmoid layer before the loss.

The loss already handles the sigmoid calculation internally.

```python
logits = torch.tensor([2.0, -2.0, 0.0])
targets = torch.tensor([1.0, 0.0, 1.0])

loss_fn = torch.nn.BCEWithLogitsLoss()
loss = loss_fn(logits, targets)

probabilities = torch.sigmoid(logits)

print("Logits:", logits)
print("Probabilities:", probabilities)
print("BCE with logits:", loss.item())
```

### Interpretation

The logits are converted to probabilities using sigmoid only for understanding the output.

The loss itself receives the raw logits.

This is the standard pattern:

**Model → raw logits → BCEWithLogitsLoss**

# Part D – Multiclass Classification

Multiclass classification means there are more than two possible classes.

Examples:
- cat / dog / horse
- red / green / blue
- low / medium / high
- classifying an image into one of several categories

## 9. Cross Entropy Loss

### Definition

Cross Entropy Loss is commonly used for multiclass classification.

A neural network usually produces one raw score called a **logit** for each class.

For example:

`[2.5, 0.5, -1.0]`

These logits represent the model's relative confidence before conversion to probabilities.

Cross entropy compares the model's class probabilities with the correct class.

### Mathematical intuition

Suppose there are three classes:

- Class 0
- Class 1
- Class 2

If the true class is Class 0, the model should assign high probability to Class 0.

If it assigns high probability to the wrong class, the loss becomes large.

### Real-world example

An image classifier predicts:

`cat = 0.70`

`dog = 0.20`

`horse = 0.10`

If the actual image is a cat, the loss is relatively low.

If the actual image is a horse, the model gave the correct class only 0.10 probability, so the loss is much larger.

### When to use

Use cross entropy for multiclass classification where each example belongs to one class.

### When not to use

It is not the direct choice for ordinary continuous regression.

For binary classification, BCE-based losses are commonly used.

### Advantages

- Standard loss for multiclass neural-network classification
- Works directly with logits in PyTorch
- Strongly penalizes confident incorrect predictions

### Limitations

- Sensitive to incorrect confident predictions
- Assumes the target representation matches the expected classification setup

### Cross Entropy – Simple numerical example

Suppose the correct class has predicted probability:

`p(correct) = 0.8`

For that example, the simplified cross-entropy contribution is:

`-log(0.8) ≈ 0.223`

If the correct class probability were only `0.1`:

`-log(0.1) ≈ 2.303`

So the loss increases greatly when the model is confident but wrong.

```python
import torch.nn.functional as F

logits = torch.tensor([[2.0, 0.5, -1.0]])
target = torch.tensor([0])

loss = F.cross_entropy(logits, target)
probabilities = torch.softmax(logits, dim=1)

print("Class probabilities:", probabilities)
print("Cross Entropy Loss:", loss.item())
```

### Interpretation

`cross_entropy` expects:
- raw logits with shape such as `(batch_size, number_of_classes)`
- class-index targets such as `0`, `1`, `2`

PyTorch internally applies the appropriate log-softmax calculation as part of the loss.

Therefore, in a normal classification model, we generally pass **raw logits** to `CrossEntropyLoss` rather than applying softmax first.

# Part E – Classification Example Using the Supplied Dataset

The supplied Online Retail dataset is naturally transactional rather than a ready-made neural-network classification dataset.

For a simple learning demonstration, I will create a binary target:

**HighValue = 1** when `TotalAmount` is above the median transaction amount, otherwise `0`.

This lets us demonstrate binary loss functions using the supplied data without pretending that this is a production classification problem.

```python
classification_df = pd.read_csv("data.csv", encoding="latin1")

classification_df = classification_df[["Quantity", "UnitPrice"]].copy()
classification_df = classification_df.replace([np.inf, -np.inf], np.nan).dropna()
classification_df = classification_df[
    (classification_df["Quantity"] > 0) &
    (classification_df["UnitPrice"] > 0)
]

classification_df["TotalAmount"] = classification_df["Quantity"] * classification_df["UnitPrice"]
threshold_value = classification_df["TotalAmount"].median()
classification_df["HighValue"] = (classification_df["TotalAmount"] > threshold_value).astype(int)

classification_df.head()
```

```python
sample_classification = classification_df.head(20).copy()

probabilities = np.clip(
    sample_classification["TotalAmount"].to_numpy() / (2 * threshold_value),
    0.01,
    0.99
)

targets = sample_classification["HighValue"].to_numpy()

binary_loss = -np.mean(
    targets * np.log(probabilities) +
    (1 - targets) * np.log(1 - probabilities)
)

print("Binary Cross Entropy on sample:", binary_loss)
```

### Interpretation

This demonstration creates a probability from transaction value only for learning purposes.

It shows how binary cross entropy evaluates predicted probabilities against binary labels.

It is **not a trained classifier**. In an actual ML workflow, a model would learn the relationship between input features and the target.

# Part F – Comparing the Loss Functions

## Regression

### MAE
Measures average absolute error.

### MSE
Squares errors, so large errors receive more penalty.

### RMSE
Takes the square root of MSE so the result returns to the target's original unit.

## Classification

### Binary Cross Entropy
Compares binary labels with predicted probabilities.

### BCEWithLogitsLoss
Combines sigmoid and BCE in a numerically stable form and expects raw logits.

### Cross Entropy Loss
Used for multiclass classification and normally receives raw logits in PyTorch.

# 10. Why Does the Model Need a Loss Function?

Imagine a student taking a test.

The student gives an answer, but without knowing how far the answer is from the correct answer, the student does not know what needs improvement.

A loss function provides this feedback to the model.

### In a neural network

**Input**
↓
**Forward propagation**
↓
**Prediction**
↓
**Loss function**
↓
**Loss value**
↓
**Backpropagation**
↓
**Gradients**
↓
**Optimizer updates weights**
↓
**Next prediction**

This is why loss is central to learning.

# 11. Loss vs Gradient vs Optimizer

These three ideas are related but different.

### Loss function
Answers:

**"How wrong is the prediction?"**

### Gradient
Answers:

**"How should each parameter change to reduce the loss?"**

### Optimizer
Uses the gradients to update the parameters.

For example:

**Loss → gradients → optimizer → updated weights**

The loss function itself does not directly change the weights.
