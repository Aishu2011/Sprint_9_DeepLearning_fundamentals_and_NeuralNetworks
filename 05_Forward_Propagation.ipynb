# 05 – Forward Propagation

# 1. What is forward propagation?

**Forward propagation** is the process of passing input data from the input layer through the neural-network layers until an output or prediction is produced.

### My understanding

I think of forward propagation as a journey.

The input enters the network, each layer processes it, and the processed information moves forward until the final layer produces the prediction.

### Real-world example

Suppose I want to predict whether a transaction belongs to a high-value category.

The input might contain:
- Quantity
- Unit price

The hidden layer combines these values and learns patterns.

The output layer produces the final prediction.

### AI/ML usage

Every normal prediction made by a feed-forward neural network requires a forward pass through the network.

# 2. Input layer

The **input layer** receives the features given to the neural network.

For example:

`x = [x1, x2]`

If I have two features:
- `x1 = Quantity`
- `x2 = UnitPrice`

then these values are the input to the network.

### Important point

The input layer usually does not perform a learned transformation by itself. It represents the features entering the network.

### AI/ML usage

The number of input features determines the number of input values expected by the first trainable layer.

# 3. Hidden layer

A **hidden layer** is a layer between the input and output.

A neuron in the hidden layer calculates a weighted sum:

`z = w1x1 + w2x2 + b`

Then an activation function is applied:

`a = activation(z)`

### My understanding

The hidden layer creates intermediate representations from the original input.

For example, instead of directly using Quantity and UnitPrice, a hidden neuron may learn a combination of them that is useful for the final prediction.

### Real-world analogy

Imagine a decision process with an intermediate stage.

Input facts → intermediate reasoning signals → final decision.

### AI/ML usage

Hidden layers allow neural networks to learn combinations and nonlinear patterns in the input data.

# 4. Output layer

The **output layer** produces the final model output.

For a binary classification problem, the network can have one output neuron.

For example:

`output = sigmoid(z_output)`

The result can be interpreted as a probability-like value for class `1`.

### For multi-class classification

The output layer can contain multiple neurons and softmax can convert their scores into class probabilities.

### AI/ML usage

The output layer is designed according to the prediction task:
- Binary classification → often one sigmoid output
- Multi-class classification → often multiple outputs with softmax
- Regression → often a linear output

# 5. Weights and biases

Every trainable neuron has **weights** and usually a **bias**.

For two inputs:

`z = w1x1 + w2x2 + b`

### Weight

A weight controls the contribution of an input.

### Bias

A bias shifts the weighted sum and gives the neuron additional flexibility.

### My understanding

Weights are like importance values, while bias is like a starting adjustment.

### AI/ML usage

Weights and biases are **parameters**. During training, the model learns values for these parameters by minimizing a loss function.

# 6. Linear transformation

Before activation, a neural-network layer performs a linear/affine transformation:

`z = Wx + b`

Here:
- `x` = input vector
- `W` = weight matrix
- `b` = bias vector
- `z` = pre-activation values

For a single neuron:

`z = w1x1 + w2x2 + b`

### My understanding

The layer first combines the inputs mathematically. Then the activation function transforms those combined values.

### AI/ML usage

This matrix operation is one of the fundamental computations inside fully connected neural-network layers.

# 7. Activation

After the linear transformation, an activation function is applied:

`a = activation(z)`

For this notebook, I will use **ReLU** in the hidden layer.

`ReLU(z) = max(0, z)`

For the final binary output, I will use **sigmoid**.

### Why different activations?

The hidden layer needs nonlinearity so the network can learn more complex patterns.

The final sigmoid converts the final score into a value between `0` and `1`.

# 8. Prediction

The final activated output is the model's prediction.

For binary classification:

`probability = sigmoid(z_output)`

A simple threshold can then be used:

- probability >= 0.5 → class `1`
- probability < 0.5 → class `0`

### Important distinction

The probability-like value and the final class label are not exactly the same thing.

For example:

`0.82 → class 1`

if the threshold is `0.5`.

# 9. Small neural network we will calculate manually

I will use a very small network:

**2 inputs → 2 hidden neurons → 1 output neuron**

Structure:

`x1, x2`

↓ 

`Hidden neuron 1, Hidden neuron 2`

↓

`Output neuron`

For the hidden layer:

`z1 = w11x1 + w12x2 + b1`

`z2 = w21x1 + w22x2 + b2`

Then:

`a1 = ReLU(z1)`

`a2 = ReLU(z2)`

For the output:

`z3 = w31a1 + w32a2 + b3`

`prediction = sigmoid(z3)`

This is the complete forward-propagation path.

# 10. Step 1 – Choose the input

For the manual example:

`x1 = 2`

`x2 = 3`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.array([2.0, 3.0])

print("Input x:", x)
```

### Result

The network receives two input values.

These values will be passed to both hidden neurons.

# 11. Step 2 – Hidden-layer weights and biases

I will use simple values so that the calculations are easy to follow.

Hidden neuron 1:

`w1 = [0.5, 0.2]`

`b1 = 0.1`

Hidden neuron 2:

`w2 = [-0.3, 0.8]`

`b2 = -0.2`

```python
W_hidden = np.array([
    [0.5, 0.2],
    [-0.3, 0.8]
])

b_hidden = np.array([0.1, -0.2])

print("Hidden weights:")
print(W_hidden)
print("Hidden biases:", b_hidden)
```

# 12. Step 3 – Hidden-layer linear transformation

The matrix form is:

`z_hidden = W_hidden x + b_hidden`

For hidden neuron 1:

`z1 = (0.5 × 2) + (0.2 × 3) + 0.1`

`z1 = 1.7`

For hidden neuron 2:

`z2 = (-0.3 × 2) + (0.8 × 3) - 0.2`

`z2 = 1.6`

```python
z_hidden = np.dot(W_hidden, x) + b_hidden

print("Hidden pre-activation values:", z_hidden)
```

### Result interpretation

The hidden layer has transformed the original two inputs into two new values.

The values before activation are called **pre-activation values** or `z` values.

# 13. Step 4 – Apply ReLU to the hidden layer

Now:

`a_hidden = ReLU(z_hidden)`

Because both values are positive in this example, ReLU keeps them unchanged.

So approximately:

`a1 = 1.7`

`a2 = 1.6`

```python
def relu(z):
    return np.maximum(0, z)

a_hidden = relu(z_hidden)

print("Hidden activations:", a_hidden)
```

### Result interpretation

The hidden layer output is now the input to the output neuron.

This is the point where the activation introduces nonlinearity into the network.

# 14. Step 5 – Output-layer weights and bias

The output neuron receives the two hidden activations.

I will use:

`W_output = [0.4, 0.6]`

`b_output = 0.1`

So:

`z_output = (0.4 × a1) + (0.6 × a2) + 0.1`

```python
W_output = np.array([0.4, 0.6])
b_output = 0.1

z_output = np.dot(W_output, a_hidden) + b_output

print("Output pre-activation value:", z_output)
```

# 15. Step 6 – Apply sigmoid to get the prediction

The final output uses sigmoid:

`prediction = sigmoid(z_output)`

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

prediction_probability = sigmoid(z_output)
prediction_class = int(prediction_probability >= 0.5)

print("Prediction probability:", round(prediction_probability, 4))
print("Predicted class:", prediction_class)
```

### Final manual result

The input started as:

`[2, 3]`

It became hidden-layer values:

`[1.7, 1.6]`

Then the output neuron produced a final score.

Sigmoid converted that score into a probability-like value.

Finally, a threshold converted the probability into a binary class.

This is forward propagation.

### In one line

**Input → Hidden linear transformation → ReLU → Output linear transformation → Sigmoid → Prediction**

# 16. Complete NumPy implementation

Now I will put the same calculations into one simple function.

This makes the forward pass easier to reuse.

```python
def forward_pass(x, W_hidden, b_hidden, W_output, b_output):
    z_hidden = np.dot(W_hidden, x) + b_hidden
    a_hidden = relu(z_hidden)
    z_output = np.dot(W_output, a_hidden) + b_output
    prediction = sigmoid(z_output)
    return z_hidden, a_hidden, z_output, prediction

z_hidden, a_hidden, z_output, prediction = forward_pass(
    x,
    W_hidden,
    b_hidden,
    W_output,
    b_output
)

print("Hidden z:", z_hidden)
print("Hidden activation:", a_hidden)
print("Output z:", z_output)
print("Prediction:", round(float(prediction), 4))
```

### My understanding

The function follows exactly the same sequence I calculated manually.

It does not train the network. The weights and biases are already provided.

So this is a **forward pass**, not a complete training process.

# 17. Using the supplied dataset

Now I will connect forward propagation to the supplied dataset.

For this educational example, I will use two numerical columns from the dataset:

- **Quantity**
- **UnitPrice**

The data will be converted into numerical input and passed through the same type of network.

This example is intended to demonstrate the mechanics of forward propagation rather than create a production-quality prediction system.

```python
df = pd.read_csv("data.csv", encoding="latin1")

print("Dataset shape:", df.shape)
print("Selected features:", ['Quantity', 'UnitPrice'])
display(df[['Quantity', 'UnitPrice']].head())
```

```python
data = df[['Quantity', 'UnitPrice']].dropna().copy()
X = data.to_numpy(dtype=float)

feature_mean = X.mean(axis=0)
feature_std = X.std(axis=0)
X_scaled = (X - feature_mean) / np.where(feature_std == 0, 1, feature_std)

print("Input matrix shape:", X_scaled.shape)
print("First five inputs:")
print(np.round(X_scaled[:5], 3))
```

### Why standardize the dataset features?

The selected features can have very different numerical scales.

Standardization puts them on a comparable scale before I pass them through the small demonstration network.

In a real neural-network project, preprocessing would be designed according to the dataset and training pipeline.

```python
X_sample = X_scaled[:5]

for row_number, sample in enumerate(X_sample, start=1):
    z_h, a_h, z_o, pred = forward_pass(
        sample,
        W_hidden,
        b_hidden,
        W_output,
        b_output
    )

    print("Sample", row_number)
    print("Input:", np.round(sample, 3))
    print("Hidden z:", np.round(z_h, 3))
    print("Hidden activation:", np.round(a_h, 3))
    print("Output z:", round(float(z_o), 3))
    print("Prediction:", round(float(pred), 4))
    print()
```

### Dataset result interpretation

Each transaction follows the same forward-propagation path.

The original feature values are:
1. converted into numerical input,
2. transformed by the hidden layer,
3. passed through ReLU,
4. transformed by the output layer,
5. passed through sigmoid,
6. converted into a prediction probability.

The weights here are manually chosen for learning, so the output should not be treated as a meaningful business prediction.

# 18. Implementing the same network using PyTorch

Now I will implement the same architecture using PyTorch.

The network is:

**2 inputs → 2 hidden neurons → 1 output neuron**

I will use:
- `torch.nn.Linear` for the linear transformations
- `torch.relu` for the hidden activation
- `torch.sigmoid` for the final binary output

This is the same mathematical process as the NumPy example.

```python
import torch

x_torch = torch.tensor([2.0, 3.0])

W_hidden_torch = torch.tensor([
    [0.5, 0.2],
    [-0.3, 0.8]
])

b_hidden_torch = torch.tensor([0.1, -0.2])

W_output_torch = torch.tensor([0.4, 0.6])
b_output_torch = torch.tensor(0.1)

print("Input tensor:", x_torch)
```

# 19. PyTorch hidden layer

The PyTorch equivalent of:

`z_hidden = W_hidden x + b_hidden`

can be written directly with tensor operations.

```python
z_hidden_torch = torch.matmul(W_hidden_torch, x_torch) + b_hidden_torch
a_hidden_torch = torch.relu(z_hidden_torch)

print("Hidden z:", z_hidden_torch)
print("Hidden activation:", a_hidden_torch)
```

### Result

The PyTorch values should match the NumPy values because I used the same inputs, weights, biases and activation function.

This is an important idea:

**PyTorch does not change the basic mathematics. It provides tools for efficiently building and training neural networks.**

# 20. PyTorch output layer

Now the hidden activation is passed to the output neuron.

```python
z_output_torch = torch.matmul(W_output_torch, a_hidden_torch) + b_output_torch
prediction_torch = torch.sigmoid(z_output_torch)

print("Output z:", z_output_torch)
print("Prediction probability:", prediction_torch)
```

### PyTorch interpretation

The final tensor contains a value between `0` and `1`.

This can be interpreted as the model's probability-like output for the positive class in this binary classification example.

# 21. Same network using `torch.nn`

In practical deep learning, I normally define the network using PyTorch's neural-network modules.

`nn.Linear` represents:

`z = Wx + b`

The activation functions are then applied between layers.

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(2, 2),
    nn.ReLU(),
    nn.Linear(2, 1),
    nn.Sigmoid()
)

print(model)
```

### What does this architecture mean?

`nn.Linear(2, 2)`

- 2 input features
- 2 hidden neurons

`nn.ReLU()`

- applies ReLU to the hidden values

`nn.Linear(2, 1)`

- 2 hidden values
- 1 output neuron

`nn.Sigmoid()`

- converts the final score to a value between 0 and 1

So the model structure directly represents:

**2 → 2 → 1**

```python
with torch.no_grad():
    model[0].weight.copy_(W_hidden_torch)
    model[0].bias.copy_(b_hidden_torch)
    model[2].weight.copy_(W_output_torch.reshape(1, 2))
    model[2].bias.copy_(b_output_torch.reshape(1))

with torch.no_grad():
    model_output = model(x_torch.reshape(1, 2))

print("Model prediction:", model_output)
```

### Result interpretation

I manually copied the same weights and biases from the NumPy example into the PyTorch model.

Therefore, the PyTorch network performs the same forward calculation.

In a real training workflow, these parameters would normally be learned automatically rather than manually entered.
